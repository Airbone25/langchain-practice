```python
from pydantic import BaseModel,Field
```

```python
class Actor(BaseModel):
    """Details of the actor in movie"""
    name: str = Field(description="Name of the Actor")
    role: str = Field(description="Role played by the actor")

class Movie(BaseModel):
    """Details of the movie"""
    title: str = Field(description="Title of the movie")
    year: str = Field(description="Year the movie released")
    description: str = Field(description="Short and concise description of the movie")
    cast: list[Actor]
    genres: list[str]
```

```python
model = model.with_structured_output(Movie)
```

```python
model.invoke("Tell me the details about movie the social network")
```
