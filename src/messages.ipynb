```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage
```

```python
messages = [
    SystemMessage("""
    You are a experienced Python developer and have expertise in web frameworks, backend systems, and databases.
    Always provide code examples and explain your reasoning.
    Be concise but thorough in your explanations. 
"""),
    HumanMessage("Explain me what is rest API. I am a complete beginner in backend and dont know where to start?")
]
```

```python
response = model.invoke(messages)

response.text
```
